```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders_dataset[col] = pd.to_datetime(orders_dataset[col])

orders_dataset.info()
```

```python
orders_dataset.head()
```

```python
orders_dataset["purchase_year"] = orders_dataset["order_purchase_timestamp"].dt.year

orders_dataset["purchase_month"] = orders_dataset["order_purchase_timestamp"].dt.month_name()

orders_dataset["purchase_quarter"] = orders_dataset["order_purchase_timestamp"].dt.quarter

orders_dataset["purchase_day"] = orders_dataset["order_purchase_timestamp"].dt.day_name()
```
